**Q1. Student Result Manager | Beginner
Create a Student class with a public name, protected _marks, and private __grade attributes. Initialize the
student through a constructor. Implement set_marks(marks), which accepts only values from 0 to 100, along with
get_marks() and get_grade(). Assign grade A for 80 to 100, B for 60 to 79, C for 40 to 59, and F below 40.
Demonstrate both a valid and an invalid marks update. Extension: Attempt direct access to __grade and explain
the result.**

```python
class Student:
    def __init__(self, name, marks):
        self.name = name        # public: anyone can use it
        self._marks = 0         # protected: "please don't touch" (a convention)
        self.__grade = "F"      # private: hidden by Python (name mangling)
        self.set_marks(marks)

    def set_marks(self, marks):
        # Only numbers from 0 to 100 are accepted
        if isinstance(marks, bool) or not isinstance(marks, (int, float)):
            print(f"Rejected: marks must be a number, got {marks!r}")
            return False
        if marks < 0 or marks > 100:
            print(f"Rejected: {marks} is outside 0-100. Marks not changed.")
            return False

        self._marks = marks
        if marks >= 80:
            self.__grade = "A"
        elif marks >= 60:
            self.__grade = "B"
        elif marks >= 40:
            self.__grade = "C"
        else:
            self.__grade = "F"
        return True

    def get_marks(self):
        return self._marks

    def get_grade(self):
        return self.__grade


def show(student):
    print(f"{student.name} | marks: {student.get_marks()} | grade: {student.get_grade()}")


student1 = Student("Ali", 85)
student2 = Student("Sara", 55)

print("--- Initial ---")
show(student1)
show(student2)

print("--- Valid update ---")
student2.set_marks(72)
show(student2)

print("--- Invalid update ---")
student1.set_marks(150)
show(student1)

print("--- Extension: direct access to __grade ---")
try:
    print(student1.__grade)
except AttributeError as error:
    print("AttributeError:", error)
# Python renamed __grade to _Student__grade, so the plain name does not exist.
print("Name-mangled access:", student1._Student__grade)
```

**Q2. Secure Bank Account | Intermediate
Create a BankAccount class with a public account holder name, private __balance, and private __pin. Implement
deposit(amount), withdraw(amount, pin), and get_balance(pin). Reject zero or negative deposits, withdrawals
exceeding the available balance, and incorrect PINs. Demonstrate a successful deposit, a successful withdrawal,
and two rejected operations. Extension: Explain why double underscores do not provide true security.**

```python
class BankAccount:
    def __init__(self, holder_name, pin, opening_balance=0):
        if opening_balance < 0:
            raise ValueError("Opening balance cannot be negative")
        self.holder_name = holder_name   # public
        self.__pin = pin                 # private
        self.__balance = opening_balance # private

    def deposit(self, amount):
        if amount <= 0:
            print("Rejected: deposit must be greater than zero")
            return False
        self.__balance += amount
        print(f"Deposit successful: Rs. {amount}")
        return True

    def withdraw(self, amount, pin):
        if pin != self.__pin:
            print("Rejected: incorrect PIN")
            return False
        if amount <= 0:
            print("Rejected: withdrawal must be greater than zero")
            return False
        if amount > self.__balance:
            print("Rejected: insufficient balance")
            return False
        self.__balance -= amount
        print(f"Withdrawal successful: Rs. {amount}")
        return True

    def get_balance(self, pin):
        if pin != self.__pin:
            print("Rejected: incorrect PIN")
            return None
        return self.__balance


account = BankAccount("Ayesha", 1234, 5000)
print("Holder:", account.holder_name)

account.deposit(2000)
account.withdraw(1500, 1234)
print("Balance: Rs.", account.get_balance(1234))

print("--- Rejected operations ---")
account.withdraw(500, 9999)       # wrong PIN
account.withdraw(100000, 1234)    # more than the balance
account.deposit(-100)             # negative deposit

print("--- Extension: is __balance really secret? ---")
try:
    print(account.__balance)
except AttributeError as error:
    print("Direct access fails:", error)

print("Mangled name works:", account._BankAccount__balance)
account._BankAccount__balance = 999999
print("Changed from outside:", account.get_balance(1234))
```

**Q3. Employee Roles | Beginner
Create a parent class Employee that stores a name and employee ID and provides display_details(). Derive
Developer (programming language) and Manager (team size). Use super() to initialize inherited attributes.
Override display_details() in each child class to include both shared and role-specific information. Create and test
one object of each type.**

```python
class Employee:
    def __init__(self, name, emp_id):
        self.name = name
        self.emp_id = emp_id

    def display_details(self):
        print(f"Name: {self.name}")
        print(f"Employee ID: {self.emp_id}")


class Developer(Employee):
    def __init__(self, name, emp_id, language):
        super().__init__(name, emp_id)   # parent sets name and emp_id
        self.language = language

    def display_details(self):
        super().display_details()        # reuse the parent's printing
        print("Role: Developer")
        print(f"Programming language: {self.language}")


class Manager(Employee):
    def __init__(self, name, emp_id, team_size):
        super().__init__(name, emp_id)
        self.team_size = team_size

    def display_details(self):
        super().display_details()
        print("Role: Manager")
        print(f"Team size: {self.team_size}")


employee = Employee("Bilal Raza", "E001")
developer = Developer("Ali Khan", "E101", "Python")
manager = Manager("Sara Ahmed", "M201", 8)

for person in [employee, developer, manager]:
    person.display_details()
    print("-----")
```

**Q4. Vehicle Rental System | Intermediate
Create a base class Vehicle with brand, model, and daily rental rate. Implement calculate_rent(days). Derive Car,
which adds a fixed insurance fee of Rs. 1,000 per rental, and Bike, which applies a 10% discount to the base
rental amount. Override calculate_rent() and reuse the parent implementation with super(). Reject rental
durations of zero or fewer days. Test both vehicle types for a three-day rental.**

```python
class Vehicle:
    def __init__(self, brand, model, daily_rate):
        self.brand = brand
        self.model = model
        self.daily_rate = daily_rate

    def calculate_rent(self, days):
        if days <= 0:
            raise ValueError("rental days must be greater than zero")
        return self.daily_rate * days


class Car(Vehicle):
    INSURANCE_FEE = 1000

    def calculate_rent(self, days):
        base_rent = super().calculate_rent(days)   # parent also checks the days
        return base_rent + self.INSURANCE_FEE


class Bike(Vehicle):
    def calculate_rent(self, days):
        base_rent = super().calculate_rent(days)
        return base_rent * 90 / 100                # 10% discount


def show_rent(vehicle, days):
    kind = type(vehicle).__name__
    try:
        rent = vehicle.calculate_rent(days)
        print(f"{kind} {vehicle.brand} {vehicle.model}, {days} days: Rs. {rent:,.2f}")
    except ValueError as error:
        print(f"{kind} {vehicle.brand} {vehicle.model}, {days} days: Rejected - {error}")


van = Vehicle("Hiace", "Van", 2000)
car = Car("Toyota", "Corolla", 5000)
bike = Bike("Honda", "CG125", 1200)

show_rent(van, 3)
show_rent(car, 3)
show_rent(bike, 3)
print("--- Invalid durations ---")
show_rent(car, 0)
show_rent(bike, -2)
```

**Q5. Notification Dispatcher | Beginner
Create EmailNotification, SMSNotification, and PushNotification classes. Each must implement send(message)
with output specific to its channel. Write a single function notify_all(notifications, message) that iterates through
different notification objects and invokes send(). Add a WhatsAppNotification class without changing the
dispatcher. Explain where polymorphism occurs.**

```python
class EmailNotification:
    def send(self, message):
        print(f"[Email] Sending email: {message}")


class SMSNotification:
    def send(self, message):
        print(f"[SMS] Sending text message: {message}")


class PushNotification:
    def send(self, message):
        print(f"[Push] Sending push alert: {message}")


def notify_all(notifications, message):
    for notification in notifications:
        notification.send(message)     # POLYMORPHISM: same call, different behavior


print("=== Three channels ===")
channels = [EmailNotification(), SMSNotification(), PushNotification()]
notify_all(channels, "Your order has shipped")


# Added later. notify_all above is NOT changed.
class WhatsAppNotification:
    def send(self, message):
        print(f"[WhatsApp] Sending WhatsApp message: {message}")


print("=== After adding WhatsApp ===")
channels.append(WhatsAppNotification())
notify_all(channels, "Your order is out for delivery")
```

**Q6. Checkout Discount Calculator | Intermediate
Create RegularCustomer, PremiumCustomer, and VIPCustomer classes, each implementing
final_price(amount). Apply discounts of 5%, 10%, and 20%, respectively. Write print_receipt(customer, amount)
so that it works with any customer type. For a bill of Rs. 10,000, show the final amount for each type. Reject
negative bill amounts. Extension: Add GoldCustomer with a 15% discount without modifying print_receipt().**

```python
def check_amount(amount):
    if amount < 0:
        raise ValueError("bill amount cannot be negative")


class RegularCustomer:
    def final_price(self, amount):
        check_amount(amount)
        return amount - amount * 5 / 100      # 5% discount


class PremiumCustomer:
    def final_price(self, amount):
        check_amount(amount)
        return amount - amount * 10 / 100     # 10% discount


class VIPCustomer:
    def final_price(self, amount):
        check_amount(amount)
        return amount - amount * 20 / 100     # 20% discount


def print_receipt(customer, amount):
    final = customer.final_price(amount)      # works for ANY customer type
    name = type(customer).__name__
    print(f"{name:<16}| Bill: Rs. {amount:,.2f} | Final: Rs. {final:,.2f}")


print("=== Bill of Rs. 10,000 ===")
for customer in [RegularCustomer(), PremiumCustomer(), VIPCustomer()]:
    print_receipt(customer, 10000)


# Extension: added later. print_receipt is NOT changed.
class GoldCustomer:
    def final_price(self, amount):
        check_amount(amount)
        return amount - amount * 15 / 100     # 15% discount


print("=== After adding GoldCustomer ===")
print_receipt(GoldCustomer(), 10000)

print("=== Negative bill ===")
try:
    print_receipt(RegularCustomer(), -500)
except ValueError as error:
    print("Rejected:", error)
```

**Q7. Shape Calculator | Beginner
Use ABC and @abstractmethod to define an abstract Shape class with area() and perimeter(). Implement both
methods in Circle and Rectangle. Accept dimensions through constructors and use pi = 3.14159. Test a circle
with radius 5 and a rectangle with width 4 and height 6. Attempt to instantiate Shape directly and explain the
outcome.**

```python
from abc import ABC, abstractmethod

PI = 3.14159


class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass


class Circle(Shape):
    def __init__(self, radius):
        if radius <= 0:
            raise ValueError("radius must be greater than zero")
        self.radius = radius

    def area(self):
        return PI * self.radius * self.radius

    def perimeter(self):
        return 2 * PI * self.radius


class Rectangle(Shape):
    def __init__(self, width, height):
        if width <= 0 or height <= 0:
            raise ValueError("width and height must be greater than zero")
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)


circle = Circle(5)
rectangle = Rectangle(4, 6)

print("Circle (radius 5)")
print(f"Area: {circle.area():.2f}")
print(f"Perimeter: {circle.perimeter():.2f}")

print("Rectangle (4 x 6)")
print(f"Area: {rectangle.area():.2f}")
print(f"Perimeter: {rectangle.perimeter():.2f}")

print("--- Trying to create Shape directly ---")
try:
    shape = Shape()
except TypeError as error:
    print("TypeError:", error)
```

**Q8. Payment Gateway | Intermediate
Define an abstract PaymentMethod class with abstract methods pay(amount) and refund(amount). Implement
CardPayment, WalletPayment, and BankTransferPayment. Each method must display the payment method,
transaction type, and amount. Validate that transaction amounts are positive. Write process_transaction(method,
amount) to call pay() for any supported payment method. Extension: Add a new payment method without
changing process_transaction().**

```python
from abc import ABC, abstractmethod


class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

    @abstractmethod
    def refund(self, amount):
        pass

    def check_amount(self, amount):          # normal method shared by all children
        if amount <= 0:
            raise ValueError("amount must be greater than zero")


class CardPayment(PaymentMethod):
    def __init__(self, card_number):
        self.card_number = card_number

    def pay(self, amount):
        self.check_amount(amount)
        print(f"Method: Card (**** {self.card_number[-4:]}) | Type: Payment | Amount: Rs. {amount:,.2f}")

    def refund(self, amount):
        self.check_amount(amount)
        print(f"Method: Card (**** {self.card_number[-4:]}) | Type: Refund | Amount: Rs. {amount:,.2f}")


class WalletPayment(PaymentMethod):
    def __init__(self, wallet_name):
        self.wallet_name = wallet_name

    def pay(self, amount):
        self.check_amount(amount)
        print(f"Method: Wallet ({self.wallet_name}) | Type: Payment | Amount: Rs. {amount:,.2f}")

    def refund(self, amount):
        self.check_amount(amount)
        print(f"Method: Wallet ({self.wallet_name}) | Type: Refund | Amount: Rs. {amount:,.2f}")


class BankTransferPayment(PaymentMethod):
    def __init__(self, bank_name):
        self.bank_name = bank_name

    def pay(self, amount):
        self.check_amount(amount)
        print(f"Method: Bank Transfer ({self.bank_name}) | Type: Payment | Amount: Rs. {amount:,.2f}")

    def refund(self, amount):
        self.check_amount(amount)
        print(f"Method: Bank Transfer ({self.bank_name}) | Type: Refund | Amount: Rs. {amount:,.2f}")


def process_transaction(method, amount):
    if not isinstance(method, PaymentMethod):
        print("Rejected: unsupported payment method")
        return
    try:
        method.pay(amount)                   # works for ANY PaymentMethod
    except ValueError as error:
        print(f"Transaction failed: {error}")


card = CardPayment("1234567812345678")
wallet = WalletPayment("JazzCash")
bank = BankTransferPayment("HBL")

print("--- Payments ---")
for method in [card, wallet, bank]:
    process_transaction(method, 5000)

print("--- Refund ---")
card.refund(1000)

print("--- Invalid cases ---")
process_transaction(card, -200)
process_transaction("cash", 100)


# Extension: a new method, process_transaction is NOT changed.
class CashOnDeliveryPayment(PaymentMethod):
    def pay(self, amount):
        self.check_amount(amount)
        print(f"Method: Cash on Delivery | Type: Payment | Amount: Rs. {amount:,.2f}")

    def refund(self, amount):
        self.check_amount(amount)
        print(f"Method: Cash on Delivery | Type: Refund | Amount: Rs. {amount:,.2f}")


print("--- New method added ---")
process_transaction(CashOnDeliveryPayment(), 2500)
```
